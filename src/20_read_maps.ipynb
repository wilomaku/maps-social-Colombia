```python
import rasterio

import numpy as np

path_image = '../data/tiffs/2009_coverage_lclu_18-3-346_e9503de6-ae67-40cb-8de8-2465b0ed4126.tif'
```

```python
# 1. Open the raster file
with rasterio.open(path_image) as src:
    # 2. Define the target row and column indices (0-based)
    row, col = 151, 100
    
    # 3. Read the specific band (e.g., band 1)
    # Note: rasterio bands are 1-indexed, but the numpy array is 0-indexed for the first dimension if read as a single band
    band_data = src.read(1)
    
    # 4. Extract the pixel value
    pixel_value = band_data[row, col]
    
    print(f"Pixel value at row {row}, col {col}: {pixel_value}")
    
    # Optional: If you need the geographic coordinates of this pixel:
    # col_coords, row_coords = src.xy(row, col)
    # print(f"Geographic coordinates: {col_coords}, {row_coords}")   
```

```python
# Compute histogram: count of pixels per unique value

with rasterio.open(path_image) as src:
    band = src.read(1)
    values, counts = np.unique(band, return_counts=True)
    values, counts = values[1:], counts[1:]  # Exclude background value (0)
    hist = dict(zip(values.tolist(), counts.tolist()))
    print("Unique values and pixel counts:")
    for v, c in hist.items():
        print(f"Value {v}: {c} pixels")
```

```python
with rasterio.open(path_image) as src:
    band = src.read(1)
    values, counts = np.unique(band, return_counts=True)
    values, counts = values[1:], counts[1:]  # Exclude background value (0)
    total = counts.sum()
    print("Unique values and pixel percentages:")
    for v, c in zip(values.tolist(), counts.tolist()):
        pct = c / total * 100
        print(f"Value {v}: {pct:.6f}%")
```
